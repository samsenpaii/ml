`requests` is used to handle the actual HTTP request sent to the website and receive its response.
 `BeautifulSoup` is used to parse and handle the HTML response received from the website after scraping.

```python
import requests
from bs4 import BeautifulSoup
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/140.0.0.0 Safari/537.36"
    ),
    "Accept": (
        "text/html,application/xhtml+xml,application/xml;"
        "q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8"
    ),
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1",
    "Sec-Fetch-Dest": "document",
    "Sec-Fetch-Mode": "navigate",
    "Sec-Fetch-Site": "none",
    "Sec-Fetch-User": "?1",
}

url = "https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page=1"

response = requests.get(
    url,
    headers=headers
)

soup = BeautifulSoup(response.content, "lxml")

```

```python
# Creates a list containing the raw HTML and data of each company's primary information
# found on the fetched page.
comapny = soup.find_all('div' , class_ = 'companyCardWrapper__primaryInformation')
```

```python
name = []
rating = []
reviews = []
companySector = []
hqInfo = []

for i in comapny:

    name.append(i.find('h2' , class_=  'companyCardWrapper__companyName').text.strip())
    rating.append(i.find('div' , class_='rating_text').text.strip())
    reviews.append(i.find('span' , class_ = 'companyCardWrapper__companyRatingCount').text.strip())
    parts = i.find('span' , class_='companyCardWrapper__interLinking').text.strip().split(" | ")
    companySector.append(parts[0])
    hqInfo.append(parts[1])
```
